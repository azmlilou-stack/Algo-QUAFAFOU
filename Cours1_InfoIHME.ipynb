```python
def somme_n_rec(n):
    if n==0:
        return 0
    return somme_n_rec(n-1)+n
    

def moyenne_n_rec(n):
    if n == 0:
        print("Erreur : n doit etre supérieur à 0")
        return None
    return (somme_n_rec(n) / n )

n = 10
print("La moyenne des ",n, "premiers nombres est : ")
moyenne_n_rec(n)
```

```python
def moyenne_n(n):
    if n==0:
        print("Erreur : n doit etre supérieur à 0")
        return None
    somme = 0
    for i in range (1, n+1):
        somme = somme+i
    moyenne= somme / n
    return moyenne

n=10000000
print("La moyenne des ",n, "premiers nombres est : ")
moyenne_n(n)
        
```

```python
def fact(n):
    "Version itérative"
    if n==0:
        return 1
    fact=1
    for i in range(1, n + 1):
        fact = fact * i
    return fact

n = 3
print ("Le factoriel de ",n,"est : (iter) ", fact(n))


def fact_rec(n):
    if n==0:
        return 1
    return fact_rec(n-1)*n

n = 3
print ("Le factoriel de ",n,"est : (rec)", fact_rec(n))
  
```

```python
def fib_iter(n):
    "Version itérative"
    if n==0:
        return 0
    if n==1:
        return 1
    a, b=0, 1
    for i in range(2, n+1):
        a, b = b, a+b
    return b

n=10
print ("Le fibonacci de ",n,"est (rec): ", fib_iter(n))

def fib_memo(n, memo=None):
    if memo is None:
        memo={}
    if n in memo:
        return memo[n]
    if n==0:
        return 0
    if n==1:
        return 1
    memo[n] = fib_memo(n-1, memo)+fib_memo(n-2, memo)
    return memo[n]

n=10
print ("Le fibonacci de ",n,"est (memo): ", fib_memo(n))
```

```python
"Version sans le modulo"
def PlusBase2(x,y):
    somme=x+y
    reste = somme - ((x+y)//2)*2
    return reste

x=5
y=1

print ("le reste de l'addition de" ,x, y,"est :", PlusBase2(x,y)) 

"Autre méthode avec le modulo"

def PlusBase2_v2(x,y):
    z=x+y
    return z%2
    
print ("le reste de l'addition de" ,x, y,"est :", PlusBase2_v2(x,y)) 
```

```python
"Multiplication de deux nombres en base 2"

def MulBase2(x,y):
    z=x*y
    return z%2

print ("le reste de la multiplication de" ,x, y,"est :", MulBase2(x,y)) 

"Version sans le modulo"
def MulBase2_v2(x,y):
    mul=x*y
    reste = mul - ((x*y)//2)*2
    return reste

x=5
y=1

print ("le reste de l'addition de" ,x, y,"est :", MulBase2_v2(x,y)) 
```
